```python
import pandas as pd

data = {
    "message" : [
          "Win a free prize now",
        "Claim your cash reward",
        "Congratulations you won a lottery",
        "Free entry in a contest",
        "Hey buddy how are you",
        "Are we meeting today",
        "Please call me when you arrive",
        "Let's study together"
    ],
    "label" : [
          "spam",
        "spam",
        "spam",
        "spam",
        "not_spam",
        "not_spam",
        "not_spam",
        "not_spam"
    ]
}

df = pd.DataFrame(data)
print(df)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()

x = vectorizer.fit_transform(df["message"])

print("Vocabulary: ",vectorizer.get_feature_names_out())
print("\nNumerical Representation: ",x.toarray())
```

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(x,df["label"])

print("Spam Detection model trained successfully")
```

```python
new_message = ["Congratulations claim your free cash prize now"]

new_message_vector = vectorizer.transform(new_message)

prediction = model.predict(new_message_vector)

print("Message: ",new_message[0])
print("Prediction: ",prediction[0])
```

```python
new_message = ["Hey buddy are we meeting today"]

new_message_vector = vectorizer.transform(new_message)
prediction = model.predict(new_message_vector)

print("Message: ",new_message[0])
print("Prediction: ",prediction[0])
```

```python
new_message = ["Congratulations claim your free cash prize"]

new_message_vector = vectorizer.transform(new_message)
prediction = model.predict_proba(new_message_vector)

print("Message: ",new_message[0])
print("Probabilities: ",prediction[0])
```

```python
messages = [
    "Win a free prize now",
    "Hey buddy how are you",
    "Claim your cash reward",
    "Please call me when you arrive",
    "Congratulations you won a lottery"
]

message_vectors = vectorizer.transform(messages)
predictions = model.predict(message_vectors)

for message, prediction in zip(messages, predictions):
    print("Message:", message)
    print("Prediction:", prediction)
    print()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

x_train,x_test,y_train,y_test = train_test_split(
    df["message"],
    df["label"],
    test_size=0.25,
    random_state=42,
    stratify=df["label"]
)

model = make_pipeline(
    CountVectorizer(),
    MultinomialNB()
)

model.fit(x_train,y_train)

y_pred = model.predict(x_test)

accuracy = accuracy_score(y_test,y_pred)

print("Actual: ",y_test.to_numpy())
print("Predicted: ",y_pred)
print("Accuracy: ",accuracy)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test,y_pred,labels=["not_spam","spam"])

print(cm)
```
