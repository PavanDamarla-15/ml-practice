```python
import numpy as np
import pandas as pd

data = {
    "Study_Hours": [2, 5, np.nan, 3, 7, 4, np.nan, 6],
    "Attendance": [60, 80, 95, np.nan, 90, 75, 98, 85],
    "Marks": [45, 70, 90, 55, np.nan, 65, 95, 78],
    "Study_Method": [
        "Self Study", "Group Study", "Self Study",
        np.nan, "Group Study", "Online",
        "Self Study", np.nan
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("Misiing values in each column: ")
print(df.isnull().sum())

print("\nTotal missing values: ")
print(df.isnull().sum().sum())
```

```python
df_mean = df.copy()

df_mean["Study_Hours"] = df_mean["Study_Hours"].fillna(df_mean["Study_Hours"].mean())

print(df_mean[["Study_Hours"]])
```

```python
df_median = df.copy()

df_median["Marks"] = df_median["Marks"].fillna(df_median["Marks"].median())

print(df_median[["Marks"]])
```

```python
print("Study Method mode: ")
print(df["Study_Method"].mode())
```

```python
df_mode = df.copy()

df_mode["Study_Method"] = df_mode["Study_Method"].fillna(df_mode["Study_Method"].mode()[0])

print(df_mode[["Study_Method"]])
```

```python
df_dropped = df.dropna()

print(df_dropped)

print("Original dataset shape: ",df.shape)
print("\nAfter dropping missing rows: ",df_dropped.shape)
```

```python
df_ffill = df.copy()

df_ffill = df_ffill.ffill()

print(df_ffill)
```

```python
df_bfill = df.copy()

df_bfill = df_bfill.bfill()

print(df_bfill)
```

```python
from sklearn.impute import SimpleImputer

print("SimpleImputer library imported successfully")
```

```python
numeric_columns = ["Study_Hours","Attendance","Marks"]

mean_imputer = SimpleImputer(strategy="mean")

df_numeric = df.copy()

df_numeric[numeric_columns] = mean_imputer.fit_transform(df_numeric[numeric_columns])

print(df_numeric)
```

```python
numeric_columns = ["Study_Hours","Attendance","Marks"]

median_imputer = SimpleImputer(strategy="median")

df_numeric = df.copy()

df_numeric[numeric_columns] = median_imputer.fit_transform(df_numeric[numeric_columns])

print(df_numeric)
```

```python
categorical_imputer = SimpleImputer(strategy="most_frequent")

df_categorical = df.copy()

df_categorical[["Study_Method"]] = categorical_imputer.fit_transform(
    df_categorical[["Study_Method"]]
)

print(df_categorical)
```

```python
df_clean = df.copy()

df_clean[numeric_columns] = median_imputer.fit_transform(
    df_clean[numeric_columns]
)

df_clean[["Study_Method"]] = categorical_imputer.fit_transform(
    df_clean[["Study_Method"]]
)

print(df_clean)

print("\nMissing values after imputation:")
print(df_clean.isnull().sum())
```

```python
constant_imputer = SimpleImputer(
    strategy="constant",
    fill_value="Unknown"
)

df_constant = df.copy()

df_constant[["Study_Method"]] = constant_imputer.fit_transform(
    df_constant[["Study_Method"]]
)

print(df_constant[["Study_Method"]])
```

```python
df_final = df.copy()

numerical_missing = SimpleImputer(strategy="median")
df_final[numeric_columns] = numerical_missing.fit_transform(df_final[numeric_columns])

categorical_missing = SimpleImputer(strategy="most_frequent")
df_final[["Study_Method"]] = categorical_missing.fit_transform(df_final[["Study_Method"]])

print("Missing values after cleaning: ",df_final.isnull().sum())
print("\nTotal missing values: ",df_final.isnull().sum().sum())
```
