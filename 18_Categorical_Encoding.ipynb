```python
import numpy as np
import pandas as pd

data = {
    "Name": ["Pavan", "Mitha", "Tamanna", "Rahul", "Arjun", "Anjali"],
    "Study_Method": [
        "Self Study",
        "Group Study",
        "Online",
        "Self Study",
        "Group Study",
        "Online"
    ],
    "Performance": [
        "Average",
        "Good",
        "Excellent",
        "Average",
        "Good",
        "Excellent"
    ],
    "Study_Hours": [2, 4, 6, 3, 5, 7]
}

df = pd.DataFrame(data)

print(df)
```

```python
categorical_columns = df.select_dtypes(include="object").columns

numerical_columns = df.select_dtypes(include=["int64","float64"]).columns

print("Categorical columns: ")
print(categorical_columns)

print("\nNumerical columns: ")
print(numerical_columns)
```

```python
from sklearn.preprocessing import LabelEncoder

df_label = df.copy()

encoder = LabelEncoder()

df_label["Study_Method_Encoded"] = encoder.fit_transform(df_label["Study_Method"])

print(df_label[["Study_Method","Study_Method_Encoded"]])
```

```python
print(encoder.classes_)
```

```python
df_onehot = pd.get_dummies(
    df,
    columns=["Study_Method"],
    dtype=int
)

print(df_onehot)
```

```python
from sklearn.preprocessing import OrdinalEncoder

df_ordinal = df.copy()

ordinal_encoder = OrdinalEncoder(
    categories=[["Average","Good","Excellent"]]
)

df_ordinal["Performance_Encoded"] = ordinal_encoder.fit_transform(
    df_ordinal[["Performance"]]
).astype(int)

print(df_ordinal[["Performance","Performance_Encoded"]])
```

```python
new_data = pd.DataFrame({
    "Study_Method": ["Library Study"]
})

print(encoder.transform(new_data["Study_Method"]))
```

```python
from sklearn.preprocessing import OneHotEncoder

onehot_encoder = OneHotEncoder(handle_unknown="ignore",sparse_output=False)

encoded_train = onehot_encoder.fit_transform(df[["Study_Method"]])

print("Training encoded data: ")
print(encoded_train)

new_encoded = onehot_encoder.transform(
    pd.DataFrame({ "Study_Method" : ["Library Study"] })
)

print("\nNew encoded data: ")
print(new_encoded)
```

```python
df_preprocessed = df.copy()

print(df_preprocessed)
```

```python
from sklearn.preprocessing import OrdinalEncoder

performance_encoder = OrdinalEncoder(
    categories=[["Average", "Good", "Excellent"]]
)

df_preprocessed["Performance"] = performance_encoder.fit_transform(
    df_preprocessed[["Performance"]]
).astype(int)

print(df_preprocessed)
```

```python
df_preprocessed = pd.get_dummies(
    df_preprocessed,
    columns=["Study_Method"],
    dtype=int
)

print(df_preprocessed)
```
