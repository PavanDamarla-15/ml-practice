```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

print("Libraries imported Successfully")
```

```python

# Create a dataset with 3 features
X = np.array([
    [2, 4, 6],
    [3, 5, 7],
    [4, 6, 8],
    [5, 8, 10],
    [6, 9, 11],
    [7, 10, 13],
    [8, 12, 14],
    [9, 13, 16],
    [10, 14, 17],
    [11, 16, 19]
])

print("Dataset shape:", X.shape)
print("\nDataset:")
print(X)
```

```python
#Standardize the Dataset
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Original shape: ",X.shape)
print("Scaled shape: ",X_scaled.shape)

print("\nScaled Dataset (first 5 rows): ",np.round(X_scaled[:5], 2))

print("\nFeature means after scaling: ",np.round(X_scaled.mean(axis=0), 5))
```

```python
#Reduce 3 features to 2 principal components
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("Original shape: ",X_scaled.shape)
print("PCA shape: ",X_pca.shape)

print("Transformed data (first 5 rows): ",np.round(X_pca[:5], 2))
```

```python
print("Explained variance ratio:")
print(np.round(pca.explained_variance_ratio_, 4))

print("\nTotal variance retained:")
print(round(pca.explained_variance_ratio_.sum(), 4))
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(X_pca[:, 0], X_pca[:, 1], s=100)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA: 3 features reduced to 2 components")

plt.grid(True)
plt.show()
```
