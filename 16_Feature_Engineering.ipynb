Day 16 — Feature Engineering
Objective: Learn how to prepare and transform raw data into meaningful features for Machine Learning models.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)
print("Feature Engineering environment is ready!")
```

```python
data = {
    "Study_Hours": [2, 5, 8, 3, 7, 4, 9, 6],
    "Attendance": [60, 80, 95, 70, 90, 75, 98, 85],
    "Previous_Marks": [45, 70, 90, 55, 85, 65, 95, 78],
    "Study_Method": [
        "Self Study",
        "Group Study",
        "Self Study",
        "Online",
        "Group Study",
        "Online",
        "Self Study",
        "Group Study"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Study_Efficiency"] = (df["Previous_Marks"] / df["Study_Hours"])

print(df[["Previous_Marks","Study_Hours","Study_Efficiency"]])
```

```python
def performance_catergory(marks):
    if marks >= 85:
        return "Excellent"
    elif marks >= 70:
        return "Good"
    else:
        return "Needs Improvement"

df["Performance_Level"] = df["Previous_Marks"].apply(performance_catergory)

print(df[["Previous_Marks","Performance_Level"]])
```

```python
df_missing = df.copy()

df_missing.loc[2, "Attendance"] = np.nan

print(df_missing.isnull().sum())
```

```python
df_missing["Attendance"] = df_missing["Attendance"].fillna(df_missing["Attendance"].median())

print(df_missing[["Attendance"]])
print("\nMissing values: ")
print(df_missing.isnull().sum())
```

```python
print(df_missing["Study_Method"].unique())
```

```python
df_encoded = pd.get_dummies(
    df_missing,
    columns=["Study_Method"],
    dtype=int
)

print(df_encoded.head())
```

```python
performance_mapping = {
    "Needs Improvement" : 0,
    "Good" : 1,
    "Excellent" : 2
}

df_encoded["Performance_Level"] = df_encoded["Performance_Level"].map(performance_mapping)

print(df_encoded["Performance_Level"])
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

print("Feature scaling libraries imported successfully")
```

```python
numeric_features = [
    "Study_Hours",
    "Attendance",
    "Previous_Marks",
    "Study_Efficiency"
]

scaler = StandardScaler()

df_standardized = df_missing.copy()

df_standardized[numeric_features] = scaler.fit_transform(df_standardized[numeric_features])

print(df_standardized[numeric_features])
```

```python
minmax_scaler = MinMaxScaler()

df_minmax = df_missing.copy()

df_minmax[numeric_features] = minmax_scaler.fit_transform(df_minmax[numeric_features])

print(df_minmax[numeric_features])
```

```python
correlation = df_missing[numeric_features].corr()

print(correlation)
```
